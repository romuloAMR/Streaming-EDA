#### Disney+

```python
!curl -L -o spreadsheets/disney_plus_shows.zip "https://www.kaggle.com/api/v1/datasets/download/shivamb/disney-movies-and-tv-shows"
!unzip spreadsheets/disney_plus_shows.zip -d spreadsheets/
!rm -r spreadsheets/disney_plus_shows.zip
!mv spreadsheets/disney_plus_titles.csv spreadsheets/disney.csv
disney = pd.read_csv("spreadsheets/disney.csv")
disney.head()
```

#### Hulu

```python
!curl -L -o spreadsheets/hulu_shows.zip "https://www.kaggle.com/api/v1/datasets/download/shivamb/hulu-movies-and-tv-shows"
!unzip spreadsheets/hulu_shows.zip -d spreadsheets/
!rm -r spreadsheets/hulu_shows.zip
!mv spreadsheets/hulu_titles.csv spreadsheets/hulu.csv
hulu = pd.read_csv("spreadsheets/hulu.csv")
hulu.head()
```

### Join

```python
netflix["streaming"] = "Netflix"
prime["streaming"] = "Prime Video"
disney["streaming"] = "Disney+"
hulu["streaming"] = "Hulu"

streamings = [netflix, prime, disney, hulu]
data = pd.concat(streamings, ignore_index=True)
data.groupby("streaming").count()
```

### Save

```python
data.info()
```

```python
data = data.drop("show_id", axis=1)
data["type"] = data["type"].astype('category')
data['title'] = data['title'].astype('string')
data["director"] = data["director"].astype('category')

data["country"] = data["country"].astype('category')
data['date_added'] = pd.to_datetime(
    data['date_added'], 
    format='%B %d, %Y',
    errors='coerce'
)
data['release_year'] = data['release_year'].astype('int16')
data["rating"] = data["rating"].astype('category')

data["listed_in"] = data["listed_in"].astype('category')
data['description'] = data['description'].astype('string')
data["streaming"] = data["streaming"].astype('category')
```

```python
data.info()
```

```python
data.to_csv("spreadsheets/streaming.csv", index=False)
```
